### Setting the directory for documents

```python
datasource = f'/Volumes/workspace/genai_course/dataset/docs'
print(datasource)
```

### We now read the files in the directory using spark.read. We load the files in the directory using load. We can also add extra columns like timestamp for debugging purpose using withColumn.


```python
from pyspark.sql.functions import current_timestamp

docs_raw_df = (spark.read.format("binaryfile")
                .load(datasource)
                .withColumn("ingestion_time", current_timestamp()))

display(docs_raw_df)
```

### We save the full dataframe as table without any transformations. This is our source of truth

```python
docs_raw_df.write.mode("overwrite").saveAsTable("bronze_docs_raw")
```

#Dispplay the contents of the table

```python
bronze_df = spark.table("bronze_docs_raw")
display(bronze_df)
```
